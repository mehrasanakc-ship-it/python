```python
#string exchange
s=input("Enter a string : ")
new_string = s[1:-1] + s[0]
print("New String =" , new_string)
```

```python
#get the first 2 and last 2 characters
s=input("Enter a string : ")
new_string = s[:2] + s[-2:]
print("New String =" , new_string)
```

```python
#swaping the character
s1=input("Enter first string : ")
s2=input("Enter second string : ")
new_s1 = s2[0] + s1[1:]
new_s2 = s1[0] + s2[1:]
print(new_s1, new_s2)
```

```python
#replace by $
s=input("Enter a string : ")
new_string = s[0] + s[1:].replace(s[0], "$")
print("New string =", new_string)
```
