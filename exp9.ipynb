```python
#adding ing or ly to a string 
s = input("Enter a string  : ")
if s.endswith("ing"):
    new_string = s + "ly"
else:
    new_string = s + "ing"
    print("New string =", new_string)
```

```python
#ckeck whether a string is a palindrome
s = input("Enter a string : ")
if s == s[::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```
