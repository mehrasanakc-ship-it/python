```python
#get the 3rd element and 3rd element from last of a tuple
t=tuple(input("Enter tuple elements separated by comma : ").split(","))
print("3rd element =", t[2])
print("3rd element from last =", t[-3])
```

```python
#create a tuple and find the average of numbers
numbers=input("Enter numbers separated by comma : ")
num_tuple = tuple(int(x) for x in numbers.split(","))
average=sum(num_tuple) / len(num_tuple)
print("Tuple =", num_tuple)
print("Average =", average)
```
