```python
#n+nn+nnn
n = int(input("Enter a number: "))

nn = n * 11
nnn = n * 111

result = n + nn + nnn

print(n, "+", nn, "+", nnn, "=", result)
```

```python
#area of the circle
import math
r = float(input("enter the radius :"))
area = math.pi * r * r
print("Area of the circle=" ,round(area, 2)) 
```

```python
#find the max number
a = int(input("enter first number :"))
b = int(input("enter second number :"))
c = int(input("enter third number :"))
biggest = max(a, b, c)
print("biggest number =" ,biggest)
```
