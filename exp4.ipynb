```python
#list of colors and display first and last colors
colors = input("Enter colors separated by comma : ")
color_list = colors.split(",")
print("First color =", color_list[0])
print("Last color =", color_list[-1])
```

```python
#accept a filename and print its extension 
filename = input("Enter a filename : ")
extension = filename.split(".")[-1]
print("Extension =", extension)
```

```python
#list of numbers and find their sum
numbers=input("Enter numbers separated by comma :")
num_list=numbers.split(",")
num_list=[int(x) for x in num_list]
total=sum(num_list)
print("list =", num_list)
print("sum =", total)
```
