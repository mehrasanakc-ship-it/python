```python
#return the length of the longest word

def longest_word(words):
    longest = max(words, key=len)
    return len(longest)
words = input("Enter words separated by space: ").split()
print("Length of longest word:" , longest_word(words))
```

```python
#pattern using nested loop

def pattern():
    for i in range(1, 6):
        for j in range(i):
            print("*", end=" ")
        print()
    for i in range(4, 0, -1):
        for j in range(i):
            print("*", end=" ")
        print()
pattern()
    
```

```python
#factors of a number

def factors(n):
    for i in range(1, n + 1):
        if n % i == 0:
            print(i, end=" ")
n = int(input("Enter a number: "))
factors(n)
```

```python
#lambda function to find the area of a square,rectangle,triangle

square = lambda a: a * a
rectangle = lambda l, b: l * b
triangle = lambda b, h: 0.5 * b * h
a = float(input("Enter side of square: "))
l = float(input("Enter length of retangle: "))
b = float(input("Enter breadth of retangle: "))
base = float(input("Enter base of triangle: "))
h = float(input("Enter height of triangle: "))
print("Area of square:", square(a))
print("Area of rectangle:", rectangle(l, b))
print("Area of triangle:", triangle(base, h))
```
