```python
# Accept 5 students' details
students = {}

roll1 = int(input("Enter roll number : "))
name1 = input("Enter name : ")
students[roll1] = name1

roll2 = int(input("Enter roll number : "))
name2 = input("Enter name : ")
students[roll2] = name2

roll3 = int(input("Enter roll number : "))
name3 = input("Enter name : ")
students[roll3] = name3

roll4 = int(input("Enter roll number : "))
name4 = input("Enter name : ")
students[roll4] = name4

roll5 = int(input("Enter roll number : "))
name5 = input("Enter name : ")
students[roll5] = name5

print("Original Dictionary:", students)

# Ascending order
ascending = dict(sorted(students.items()))

print("Ascending order:", ascending)

# Descending order
descending = dict(sorted(students.items(), reverse=True))

print("Descending order:", descending)
```

```python
#merge two dictionaries using update
dic1 = {}
roll1 = int(input("Enter roll numbers : "))
name1 = input("Enter name :")
dic1[roll1] = name1

roll2 = int(input("Enter roll numbers : "))
name2 = input("Enter name :")
dic1[roll2] = name2

roll3 = int(input("Enter roll numbers : "))
name3 = input("Enter name :")
dic1[roll3] = name3

dic2 = {}
roll4 = int(input("Enter roll numbers : "))
name4 = input("Enter name :")
dic2[roll4] = name4

roll5 = int(input("Enter roll numbers : "))
name5 = input("Enter name :")
dic2[roll5] = name5

print("Dictionary 1:", dic1)
print("Dictionary 2:", dic2)
dic1.update(dic2)
print("Merged Dictionary:", dic1)

```
