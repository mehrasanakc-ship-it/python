```python
#clr in set 1 but not in set 2
set1=set(input("Enter colors in set 1 : ").split())

set2=set(input("Enter colors in set 2 : ").split())

print("set 1:", set1)

print("set 2:", set2)

result = set1 - set2

print("Colors in set 1 but not in set 2: ",result)
```

```python
#cls in set 1 and 2
set1=set(input("Enter colors in set 1 : ").split())

set2=set(input("Enter colors in set 2 : ").split())

print("set 1:", set1)

print("set 2:", set2)

result = set1 & set2

print("common colors:",result)
```
